## 1. Problem Explanation

Searching and sorting are fundamental operations in data processing. Searching is used to locate a particular value in a collection, while sorting arranges values in a chosen order, such as ascending order.

This project implements and demonstrates five searching and sorting algorithms:

- **Linear Search** checks values one at a time and works with sorted or unsorted lists.
- **Binary Search** repeatedly examines the middle of a sorted list and eliminates half of the remaining search range.
- **Bubble Sort** compares adjacent values and swaps them when they are in the wrong order.
- **Selection Sort** finds the smallest value in the unsorted portion and places it in the next correct position.
- **Insertion Sort** builds a sorted portion by inserting each next value into its correct position.

A sample list of numeric values is used to demonstrate these operations, verify results through tests, and analyse algorithm time complexity.

## 2. Data Structure and Sample Data

The main data structure is a **Python list**. A list stores multiple values in an ordered collection and allows access to values by index. It is suitable for demonstrating all five algorithms in this project.

```python
# Sample data
sample_numbers = [34, 7, 23, 32, 5, 62, 14, 7]

print("Original sample data:", sample_numbers)
print("Number of elements:", len(sample_numbers))
print("Target to find:", 23)
print("Target not in the list:", 99)
```

## 3. Python Implementation: Searching Algorithms

### 3.1 Linear Search

Linear Search starts at the beginning of the list and compares each item with the target. It returns the index of the first match, or `-1` if the target is not found. It does not require the list to be sorted.

```python
def linear_search(numbers, target):
    """Return the index of the first target occurrence, or -1 if absent."""
    for i in range(len(numbers)):
        if numbers[i] == target:
            return i
    return -1


# Demonstration
linear_result = linear_search(sample_numbers, 23)
missing_linear_result = linear_search(sample_numbers, 99)

print("Linear Search for 23:", linear_result)
print("Linear Search for 99:", missing_linear_result)
```

**Algorithm steps**
1. Begin at index `0`.
2. Compare the current element with the target.
3. If they match, return the current index.
4. Otherwise, continue to the next element.
5. If the end of the list is reached without a match, return `-1`.

**Complexity:** Best case `O(1)` when the first element matches; average and worst cases `O(n)` because the algorithm may inspect a number of elements proportional to the list length. Auxiliary space is `O(1)`. 

### 3.2 Binary Search

Binary Search uses a divide-and-conquer approach. It checks the middle element and narrows the search range to the left or right half. **The input list must be sorted** before Binary Search is used.

```python
def binary_search(arr, target):
    """Search a sorted list and return the target index, or -1 if absent."""
    low = 0
    high = len(arr) - 1

    while low <= high:
        mid = (low + high) // 2

        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1

    return -1


sorted_numbers_for_search = sorted(sample_numbers)
binary_result = binary_search(sorted_numbers_for_search, 23)
missing_binary_result = binary_search(sorted_numbers_for_search, 99)

print("Sorted data for Binary Search:", sorted_numbers_for_search)
print("Binary Search for 23:", binary_result)
print("Value at returned index:", sorted_numbers_for_search[binary_result] if binary_result != -1 else "Not found")
print("Binary Search for 99:", missing_binary_result)
```

**Algorithm steps**
1. Set `low` to the first index and `high` to the last index.
2. Calculate the middle index.
3. If the middle value equals the target, return its index.
4. If the target is greater, continue in the right half; otherwise, continue in the left half.
5. Repeat while `low <= high`; return `-1` if no match is found.

**Complexity:** Best case `O(1)` when the middle element is the target. Average and worst cases are `O(log n)` because the remaining search range is halved at each step. Auxiliary space is `O(1)` for this iterative implementation.

## 4. Python Implementation: Sorting Algorithms

Each sorting function below returns the list in ascending order. A copy of the sample data is passed to each function so the original list remains unchanged.

### 4.1 Bubble Sort

Bubble Sort repeatedly compares adjacent elements and swaps them when the left element is greater than the right element. The largest unsorted value moves toward the end during each pass. The implementation stops early if a pass makes no swaps.

```python
def bubble_sort(arr):
    """Sort a list in ascending order using Bubble Sort."""
    arr = arr.copy()
    n = len(arr)

    for i in range(n):
        swapped = False
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True
        if not swapped:
            break

    return arr


print("Bubble Sort:", bubble_sort(sample_numbers))
```

**Complexity:** Best case `O(n)` when the list is already sorted, due to the early-stop check. Average and worst cases are `O(n²)`. Auxiliary space is `O(n)` in this version because it makes a copy to preserve the original list; the sorting process itself uses `O(1)` extra space.

### 4.2 Selection Sort

Selection Sort divides the list into a sorted portion and an unsorted portion. On each pass, it finds the smallest value in the unsorted portion and swaps it into the next position.

```python
def selection_sort(arr):
    """Sort a list in ascending order using Selection Sort."""
    arr = arr.copy()
    n = len(arr)

    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
        arr[i], arr[min_idx] = arr[min_idx], arr[i]

    return arr


print("Selection Sort:", selection_sort(sample_numbers))
```

**Complexity:** Best, average, and worst cases are `O(n²)` because the algorithm scans the remaining unsorted portion on each pass. Auxiliary space is `O(n)` in this version due to copying the input; the selection-sort operations themselves use `O(1)` extra space.

### 4.3 Insertion Sort

Insertion Sort considers the first element sorted, then takes each subsequent value and inserts it into its correct place in the sorted portion by shifting larger values to the right.

```python
def insertion_sort(arr):
    """Sort a list in ascending order using Insertion Sort."""
    arr = arr.copy()

    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1

        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1

        arr[j + 1] = key

    return arr


print("Insertion Sort:", insertion_sort(sample_numbers))
```

**Complexity:** Best case `O(n)` when the list is already sorted, because only one comparison is needed for each next element. Average and worst cases are `O(n²)`. Auxiliary space is `O(n)` in this version because a copy is made; the insertion-sort operations themselves use `O(1)` extra space.
